### Значение аргументов по умолчанию

```python
def relative_difference(x: float, y: float, verbose: bool = False) -> float | None:
    delta = x - y
    if verbose:
        print(f'Delta: {delta}')
    mean = (x + y) / 2
    if verbose:
        print(f'Mean: {mean}')
    if abs(mean) < 1e-10:
        if verbose:
            print('Mean is equal to zero!')
        return None
    return abs(delta / mean)
```

### Опциональные аргументы

```python
from collections.abc import Callable

def relative_difference(
        x: float, y: float,
        normalize_by: Callable[[float, float], float] | None = None
    ) -> float | None:

    delta = x - y

    if normalize_by is not None:
        norm = normalize_by(x, y)
    else:
        norm = 1

    if abs(norm) < 1e-10:
        return None
    return abs(delta / norm)
```

### Исключительно именованные аргументы

```python
def relative_difference(
        x: float, y: float, *,
        normalize_by: Callable[[float, float], float] | None = None
    ) -> float | None:

    delta = x - y

    if normalize_by is not None:
        norm = normalize_by(x, y)
    else:
        norm = 1.0

    if abs(norm) < 1e-10:
        return None
    return abs(delta / norm)
```

### Исключительно позиционные аргументы

```python
def relative_difference(
        x: float, y: float,
        /,
        *,
        normalize_by: Callable[[float, float], float] | None = None
    ) -> float | None:

    delta = x - y

    if normalize_by is not None:
        norm = normalize_by(x, y)
    else:
        norm = 1.0

    if abs(norm) < 1e-10:
        return None
    return abs(delta / norm)
```

### А если None нужен для другого?

```python
class _Missing:
    pass


MISSING = _Missing()


def function(list_argument: list[str] | None | _Missing = MISSING) -> None:
    if list_argument is None:
        return

    the_list: tp.List[str]
    if isinstance(list_argument, _Missing):
        the_list = []
    else:
        the_list = list_argument

    the_list.append("Hi!")
    print(the_list)
```

```python
function()
```

```python
print(function(None))
```

```python
messages: tp.List[str] = ["Hello!"]
function(messages)
print(messages)  # ['Hello!', 'Hi!']
```

<b>Hint:</b> Если вам нужно отличать случай, когда вам передали "None", от дефолтного значения аргумента,

В ветке else проверка isinstance исключает _Missing: там list_argument имеет тип tp.List[str].
В этом примере любой экземпляр _Missing означает, что аргумент не задан.

### Функция от произвольного числа аргументов

$$
\sqrt{\frac{1}{n}\sum_{i=1}^{n} x_i^2}
$$

```python
def root_mean_square(args: list[float]) -> float:
    if not args:
        return 0.0

    squares_sum = sum(x ** 2 for x in args)

    mean = squares_sum / len(args)
    return mean ** 0.5
```

```python
root_mean_square([4, 8, 15, 16, 23, 42])
```

## *args

```python
def root_mean_square(*args: float) -> float:
    if not args:
        return 0.0

    squares_sum = sum(x ** 2 for x in args)

    mean = squares_sum / len(args)
    return mean ** 0.5
```

<div class="alert alert-warning">
<b>Типизация:</b> В записи *args: float аннотация относится к каждому аргументу. Внутри функции args — кортеж типа tuple[float, ...].
</div>

```python
root_mean_square(4, 8, 15, 16, 23, 42)
```

## **kwargs

```python
def root_mean_square(*args: float, **kwargs: tp.Any) -> float:
    verbose = kwargs.get('verbose', False)

    if not len(args):
        if verbose:
            print('Empty arguments list!')
        return 0.0

    squares_sum = sum(x ** 2 for x in args)
    if verbose:
        print(f'Sum of squares: {squares_sum}')

    mean = squares_sum / len(args)
    if verbose:
        print(f'Mean square: {mean}')

    return mean ** 0.5
```

```python
root_mean_square(4, 8, 15, 16, 23, 42, verbose=True)
```

```python
root_mean_square(verbose=True)
```

```python
# Видите ли подвох?

root_mean_square(3, 4, verbsoe=True)
```

### Частичное применение: partial и Placeholder

partial создаёт вызываемый объект с заранее заданными аргументами.
Placeholder (Python 3.14+) оставляет место для позиционного аргумента, который будет передан при вызове.

```python
from functools import partial

# Фиксируем первый аргумент pow — основание степени.
power_of_two = partial(pow, 2)

print(power_of_two(3))
print(power_of_two(5))
```

```python
from functools import Placeholder

square = partial(pow, Placeholder, 2)

print(square(3))
print(square(5))
```

#### T-строки: Python 3.14

F-строка сразу создаёт готовую строку. T-строка создаёт объект Template: постоянные части текста, значения подстановок и правила форматирования хранятся отдельно.

```python
name = 'Анна'
price = 123.456

message = f"Для {name}: {price:.2f} ₽"
template = t"Для {name}: {price:.2f} ₽"

print(message)
print(type(template))
```

```python
print(template.strings)
print(template.values)
```

```python
item = template.interpolations[1]

print(item.value)
print(item.expression)
print(item.format_spec)
```

<div class="alert alert-info">
<b>Hint:</b> Значение price осталось числом, а правило .2f хранится отдельно. Обработчик может собрать текст, экранировать подстановки для HTML или сохранить исходные значения в лог. Сама t-строка не выполняет такую обработку.

## Немного про UTF-8

```python

def letter_summary(letter: str) -> tp.Tuple[str, int, str, bytes, str]:
    dec_code = ord(letter)
    hex_code = hex(dec_code)
    bytes_written = letter.encode('utf-8')
    return letter, dec_code, hex_code, bytes_written, ' '.join([bin(byte)[2:].rjust(8, '0') for byte in bytes_written])
```

## Файлы

Если не указать encoding, кодировка текстового файла выбирается по настройкам окружения и режиму UTF-8. В нашем окружении это UTF-8, но на другой машине она может отличаться. Кодировка по умолчанию не определяется по содержимому файла.

```python
f = open('tail_access.log')
content = f.read()
print(len(content))
f.close()  # Важно закрыть файл после использования.
```

Читать весь файл в память может быть очень дорого

```python
f = open('tail_access.log')
print(f.readline())

# Продолжаем с текущей позиции: первая строка уже прочитана.
length = 0
for line in f:
    length += len(line)

print(length)
f.close()
```

А если там не строки? Или нужно прочитать сразу пачку строк?

```python
f = open('tail_access.log')
chunk_size = 512
lines = 0

while True:
    chunk = f.read(chunk_size)
    if not chunk:
        break
    lines += chunk.count('\n')

print(lines)
print(repr(f.read()))
f.close()
```

```python
f = open('unicode_file.txt', encoding='windows-1251')
print(f.read(5))
f.close()
```

```python
f = open('unicode_file.txt', encoding='utf-8')
print(f.read(5))
f.close()
```

А если нужно вернуться в начало прочитанного файла? Обязательно открывать заново?

```python
f = open('tail_access.log')
f.read()
print(repr(f.read()))
```

```python
import io
f.seek(0, io.SEEK_SET)
```

```python
f.readline()
```

```python
f.tell()
```

В бинарном файле tell() возвращает позицию в байтах. В текстовом — значение для восстановления позиции через seek(). Не прибавляйте к нему число символов: чтобы пропустить одну кодовую позицию, используйте read(1).

```python
position = f.tell()
line = f.readline()

f.seek(position)
f.readline() == line
```

### А если внутри with случится ошибка?

```python
try:
    with open('/tmp/junk.txt', encoding='utf-8') as file:
        raise ValueError('Ошибка при обработке файла')
except ValueError:
    print('Перехватили ошибку')
```

```python
# А тут что будет?
print(file.closed)
```

<b>Hint:</b> При выходе из with файл закрывается и при обычном завершении, и при исключении. Само исключение при этом не подавляется: здесь его перехватывает except.

## In-memory streams

### StringIO

```python
import io
```

```python
print(output.getvalue())
```

### StringIO: read() и getvalue()

```python
buffer = io.StringIO()
buffer.write('Hello!')

# Будут ли результаты одинаковыми?
print(repr(buffer.read()))
print(repr(buffer.getvalue()))
```

```python
# А теперь?
buffer.seek(0)
print(repr(buffer.read()))
buffer.close()
```

<div class="alert alert-info">
<b>Hint:</b> read() читает с текущей позиции и сдвигает её. После записи позиция находится в конце. getvalue() возвращает всё содержимое, не меняя позицию. seek(0) возвращает позицию в начало.
</div>

### BytesIO
